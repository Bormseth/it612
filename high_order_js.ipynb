```python
%%writefile code.js

// Part A
console.log("Part A");
const data = [5, 12, 3, 20, 11, 9, 30];
let result = [];
for (let i = 0; i < data.length; i++) {
  if (data[i] > 10) {
    result.push(data[i] * 2);
  }
}
console.log(result);
result = data.filter(x => x > 10).map(x => x * 2);
console.log(result);

// Part B
console.log("Part B");
const numbers = [1, 2, 3, 4, 5, 6];
let doubleSum = numbers.filter(x => x % 2 == 0)
  .map(x => x * 2)
  .reduce((acc, x) => acc + x, 0);
console.log(doubleSum);

// Part C
console.log("Part C");
const users = [
  { id: 1, name: "Alice", age: 25 },
  { id: 2, name: "Bob", age: 17 },
  { id: 3, name: "Carol", age: 19 }
];
console.log(users.some(x => x.age < 18));
console.log(users.find(x => x.id === 3));
users.forEach(x => console.log(x.name));

// Part D
console.log("Part D");
function repeat(n, action) {
  for (let i = 0; i < n; i++) {
    action(i);
  }
}
repeat(3, console.log); // 0, 1, 2
repeat(4, i => console.log("Step " + i));

```
